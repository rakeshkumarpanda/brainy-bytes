```python
# A small helper. check(tests) runs each test of a task and tells you which ones pass.
# Each test is a pair: the call to make, and the answer we expect.
import json

def check(tests):
    passed = 0
    for call, expected in tests:
        try:
            got = eval(call)
        except Exception as error:
            got = "error: " + str(error)
        same = json.dumps(got, sort_keys=True, default=repr) == json.dumps(expected, sort_keys=True, default=repr)
        if same:
            passed = passed + 1
            print("PASS ", call)
        else:
            print("FAIL ", call, " expected", expected, " but got", got)
    if passed == len(tests):
        print("All", len(tests), "tests pass. Well done!")
    else:
        print(passed, "of", len(tests), "tests pass. Keep going, or look at the solution below.")
```

### Minimax on the 27 leaf slide example

```python
# Minimax on the 27 leaf example of the slides. A tree is a list of subtrees, a leaf is a number.
leaves = [2, 1, 5, 6, 2, 11, 8, 5, 1, 3, 2, 4, 10, 11, 12, 1, 2, 1, 2, 1, 3, 6, 1, 11, 2, 5, 1]

def build(values, size):                     # cut a flat list into groups of 3, then groups of groups
    groups = []
    for i in range(0, len(values), size):
        groups.append(values[i:i + size])
    return groups

tree = build(build(leaves, 3), 3)            # A has B1, B2, B3. Each B has three C nodes. Each C has 3 leaves.

def minimax(node, is_max):
    if type(node) == int:
        return node                          # a leaf: its utility
    values = []
    for child in node:
        values.append(minimax(child, not is_max))
    if is_max:
        return max(values)
    return min(values)

for i in range(3):
    print("B" + str(i + 1), "=", minimax(tree[i], False))
print("A =", minimax(tree, True))
```

### Minimax on the slide tic-tac-toe position

```python
# Minimax on the tic-tac-toe position of the slides. O is MAX (+1 if O wins), X is MIN (-1 if X wins).
# Squares are numbered 0 to 8, row by row. " " is an empty square.
start = ["O", "X", "O",
         "X", " ", "X",
         " ", "O", " "]
lines = [(0, 1, 2), (3, 4, 5), (6, 7, 8), (0, 3, 6), (1, 4, 7), (2, 5, 8), (0, 4, 8), (2, 4, 6)]

def winner(b):
    for a, c, d in lines:
        if b[a] != " " and b[a] == b[c] == b[d]:
            return b[a]
    return None

def minimax(b, player):
    w = winner(b)
    if w == "O":
        return 1
    if w == "X":
        return -1
    if " " not in b:
        return 0                                   # a draw
    values = []
    for i in range(9):
        if b[i] == " ":
            b[i] = player
            values.append(minimax(b, "X" if player == "O" else "O"))
            b[i] = " "                             # undo the move
    return max(values) if player == "O" else min(values)

for i in range(9):
    if start[i] == " ":
        start[i] = "O"
        print("O plays square", i, "-> value", minimax(start, "X"))
        start[i] = " "
```

### Solution to task 1: Minimax on a nested list tree

The second test is the classic AIMA figure 5.2 tree: MIN nodes 3, 2, 2, root 3.

```python
def minimax(node, is_max):
    if type(node) == int:
        return node
    values = []
    for child in node:
        values.append(minimax(child, not is_max))
    if is_max:
        return max(values)
    return min(values)
```

### Solution to task 2: Best tic-tac-toe move

In the second test X wins at once by completing the top row at square 2. In the third test O must block the diagonal 2, 4, 6 at square 6, because any other move lets X win next turn.

```python
def best_move(board, player):
    lines = [(0, 1, 2), (3, 4, 5), (6, 7, 8), (0, 3, 6), (1, 4, 7), (2, 5, 8), (0, 4, 8), (2, 4, 6)]

    def winner(b):
        for a, c, d in lines:
            if b[a] != " " and b[a] == b[c] == b[d]:
                return b[a]
        return None

    def value(b, turn):
        w = winner(b)
        if w == "O":
            return 1
        if w == "X":
            return -1
        if " " not in b:
            return 0
        results = []
        for i in range(9):
            if b[i] == " ":
                b[i] = turn
                results.append(value(b, "X" if turn == "O" else "O"))
                b[i] = " "
        if turn == "O":
            return max(results)
        return min(results)

    b = list(board)
    best = None
    best_value = None
    for i in range(9):
        if b[i] == " ":
            b[i] = player
            v = value(b, "X" if player == "O" else "O")
            b[i] = " "
            if best is None or (player == "O" and v > best_value) or (player == "X" and v < best_value):
                best = i
                best_value = v
    return best
```
