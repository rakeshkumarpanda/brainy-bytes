```python
# A small helper. check(tests) runs each test of a task and tells you which ones pass.
# Each test is a pair: the call to make, and the answer we expect.
import json

def check(tests):
    passed = 0
    for call, expected in tests:
        try:
            got = eval(call)
        except Exception as error:
            got = "error: " + str(error)
        same = json.dumps(got, sort_keys=True, default=repr) == json.dumps(expected, sort_keys=True, default=repr)
        if same:
            passed = passed + 1
            print("PASS ", call)
        else:
            print("FAIL ", call, " expected", expected, " but got", got)
    if passed == len(tests):
        print("All", len(tests), "tests pass. Well done!")
    else:
        print(passed, "of", len(tests), "tests pass. Keep going, or look at the solution below.")
```

### A* on the Romania map

```python
# A* search on the Romania map, with the straight line distance to Bucharest as h.
roads = [("Arad", "Zerind", 75), ("Arad", "Sibiu", 140), ("Arad", "Timisoara", 118), ("Zerind", "Oradea", 71),
         ("Oradea", "Sibiu", 151), ("Timisoara", "Lugoj", 111), ("Lugoj", "Mehadia", 70), ("Mehadia", "Drobeta", 75),
         ("Drobeta", "Craiova", 120), ("Craiova", "Rimnicu Vilcea", 146), ("Craiova", "Pitesti", 138),
         ("Sibiu", "Fagaras", 99), ("Sibiu", "Rimnicu Vilcea", 80), ("Rimnicu Vilcea", "Pitesti", 97),
         ("Fagaras", "Bucharest", 211), ("Pitesti", "Bucharest", 101)]
h = {"Arad": 366, "Zerind": 374, "Oradea": 380, "Sibiu": 253, "Timisoara": 329, "Lugoj": 244, "Mehadia": 241,
     "Drobeta": 242, "Craiova": 160, "Rimnicu Vilcea": 193, "Fagaras": 176, "Pitesti": 100, "Bucharest": 0}

def astar(start, goal):
    frontier = [(h[start], 0, [start])]          # (f, g, path)
    best_g = {start: 0}
    while len(frontier) > 0:
        frontier.sort()
        f, g, path = frontier.pop(0)
        city = path[-1]
        print("expand", city, " g =", g, " h =", h[city], " f =", f)
        if city == goal:
            return g, path
        for a, b, cost in roads:
            if city == a or city == b:
                nxt = b if city == a else a
                new_g = g + cost
                if nxt not in best_g or new_g < best_g[nxt]:
                    best_g[nxt] = new_g
                    frontier.append((new_g + h[nxt], new_g, path + [nxt]))
    return None

print(astar("Arad", "Bucharest"))
```

### An inconsistent heuristic that breaks graph search A*

```python
# An admissible but INCONSISTENT heuristic breaks graph search A* that never reopens a closed state.
# Roads: S-A 1, A-B 1, S-B 4, B-G 3. The best path is S, A, B, G with cost 5.
roads = {"S": [("A", 1), ("B", 4)], "A": [("B", 1)], "B": [("G", 3)], "G": []}
h = {"S": 0, "A": 4, "B": 0, "G": 0}     # h(A) = 4 > cost(A, B) + h(B) = 1, so h is not consistent

frontier = [(h["S"], 0, ["S"])]
closed = []
while len(frontier) > 0:
    frontier.sort()
    f, g, path = frontier.pop(0)
    node = path[-1]
    if node in closed:
        continue
    print("expand", node, "with g =", g, "f =", f)
    if node == "G":
        print("returned cost", g, "but the best cost is 5")
        break
    closed.append(node)
    for nxt, cost in roads[node]:
        if nxt not in closed:
            frontier.append((g + cost + h[nxt], g + cost, path + [nxt]))
```

### Solution to task 1: Write A* search

The second test is the slide 58 trap: A* pays 30, greedy would pay 110.

```python
def astar(edges, h, start, goal):
    frontier = [(h[start], 0, [start])]
    best_g = {start: 0}
    while len(frontier) > 0:
        frontier.sort()
        f, g, path = frontier.pop(0)
        node = path[-1]
        if node == goal:
            return [g, path]
        for a, b, cost in edges:
            if node == a or node == b:
                nxt = b if node == a else a
                new_g = g + cost
                if nxt not in best_g or new_g < best_g[nxt]:
                    best_g[nxt] = new_g
                    frontier.append((new_g + h[nxt], new_g, path + [nxt]))
    return None
```

### Solution to task 2: Check a heuristic for consistency

The first test is the inconsistent example of the lesson: h(A) = 4 &gt; 1 + h(B) = 1.

```python
def is_consistent(edges, h):
    for a, b, cost in edges:
        if h[a] > cost + h[b]:
            return False
        if h[b] > cost + h[a]:
            return False
    return True
```
