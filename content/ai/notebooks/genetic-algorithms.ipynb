## Setup

```python
# A small helper. check(tests) runs each test of a task and tells you which ones pass.
# Each test is a pair: the call to make, and the answer we expect.
import json

def check(tests):
    passed = 0
    for call, expected in tests:
        try:
            got = eval(call)
        except Exception as error:
            got = "error: " + str(error)
        same = json.dumps(got, sort_keys=True, default=repr) == json.dumps(expected, sort_keys=True, default=repr)
        if same:
            passed = passed + 1
            print("PASS ", call)
        else:
            print("FAIL ", call, " expected", expected, " but got", got)
    if passed == len(tests):
        print("All", len(tests), "tests pass. Well done!")
    else:
        print(passed, "of", len(tests), "tests pass. Keep going, or look at the solution below.")
```

## Examples from the chapter

### One generation of the slide x² example

```python
# One generation of the slide example: maximise f(x) = x * x where x is a 5 bit number.
population = ["01101", "11000", "01000", "10011"]

fitness = []
for s in population:
    x = int(s, 2)                       # read the bits as a whole number
    fitness.append(x * x)
total = sum(fitness)
average = total / len(fitness)
print("fitness:", fitness, " total:", total, " average:", average)

for i in range(4):
    p = fitness[i] / total              # chance of being picked by the roulette wheel
    expected = fitness[i] / average     # expected number of copies in the mating pool
    print(population[i], " p =", round(p, 2), " expected copies =", round(expected, 2))

def crossover(a, b, cut):
    return a[:cut] + b[cut:], b[:cut] + a[cut:]

# the slide mating pool: string 2 twice, string 3 dropped. Pairs and cut points from the slide.
print(crossover("01101", "11000", 4))   # -> 01100 and 11001
print(crossover("11000", "10011", 2))   # -> 11011 and 10000
```

### Roulette wheel selection

```python
# Roulette wheel selection: each string gets a slice of the wheel as big as its fitness.
import random
random.seed(1)

population = ["01101", "11000", "01000", "10011"]
fitness = [169, 576, 64, 361]

def spin():
    point = random.random() * sum(fitness)   # where the ball stops
    running = 0
    for i in range(len(fitness)):
        running = running + fitness[i]
        if point < running:
            return population[i]
    return population[-1]

counts = {}
for k in range(20):
    s = spin()
    counts[s] = counts.get(s, 0) + 1
print(counts)       # 11000 (fitness 576) should come up most often
```

### Solution to task 1: Crossover and mutation

These are the two crossovers of the slide example.

```python
def crossover(a, b, cut):
    child1 = a[:cut] + b[cut:]
    child2 = b[:cut] + a[cut:]
    return [child1, child2]

def flip(s, i):
    if s[i] == "0":
        new_bit = "1"
    else:
        new_bit = "0"
    return s[:i] + new_bit + s[i + 1:]
```

```python
check([("crossover('01101', '11000', 4)", ['01100', '11001']), ("crossover('11000', '10011', 2)", ['11011', '10000']), ("flip('01101', 0)", '11101'), ("flip('01101', 4)", '01100')])
```

### Solution to task 2: The fitness table of one generation

64 ⁄ 1170 = 0.0547, which rounds to 0.05. The slide shows 0.06, rounding up by hand.

```python
def ga_table(population):
    fitness = []
    for s in population:
        x = int(s, 2)
        fitness.append(x * x)
    total = sum(fitness)
    average = total / len(fitness)
    chances = []
    copies = []
    for f in fitness:
        chances.append(round(f / total, 2))
        copies.append(round(f / average, 2))
    return [fitness, total, chances, copies]
```

```python
check([("ga_table(['01101', '11000', '01000', '10011'])", [[169, 576, 64, 361], 1170, [0.14, 0.49, 0.05, 0.31], [0.58, 1.97, 0.22, 1.23]]), ("ga_table(['00001', '00010'])", [[1, 4], 5, [0.2, 0.8], [0.4, 1.6]])])
```
