## Setup

```python
# A small helper. check(tests) runs each test of a task and tells you which ones pass.
# Each test is a pair: the call to make, and the answer we expect.
import json

def check(tests):
    passed = 0
    for call, expected in tests:
        try:
            got = eval(call)
        except Exception as error:
            got = "error: " + str(error)
        same = json.dumps(got, sort_keys=True, default=repr) == json.dumps(expected, sort_keys=True, default=repr)
        if same:
            passed = passed + 1
            print("PASS ", call)
        else:
            print("FAIL ", call, " expected", expected, " but got", got)
    if passed == len(tests):
        print("All", len(tests), "tests pass. Well done!")
    else:
        print(passed, "of", len(tests), "tests pass. Keep going, or look at the solution below.")
```

## Examples from the chapter

### A simple reflex agent cleaning two squares

```python
# A simple reflex agent for the two square vacuum world.
# It looks only at the current percept: where it is, and whether that square is dirty.
def reflex_agent(location, status):
    if status == "Dirty":
        return "Suck"
    if location == "A":
        return "Right"
    return "Left"

# Run it for 4 time steps. Both squares start dirty.
dirt = {"A": True, "B": True}
location = "A"
score = 0
for t in range(1, 5):
    status = "Dirty" if dirt[location] else "Clean"
    action = reflex_agent(location, status)
    if action == "Suck":
        dirt[location] = False
    elif action == "Right":
        location = "B"
    else:
        location = "A"
    clean_squares = (not dirt["A"]) + (not dirt["B"])
    score = score + clean_squares
    print("time", t, "percept", [location, status], "action", action, "score", score)
```

### How fast a table driven agent grows

```python
# How big must a table driven agent be?
# It needs one row for every possible sequence of percepts.
percepts = 4        # vacuum world: 2 locations times 2 states (Dirty or Clean)
total_rows = 0
for t in range(1, 6):
    rows_for_t = percepts ** t
    total_rows = total_rows + rows_for_t
    print("lifetime", t, "steps needs", total_rows, "rows")
```

### Solution to task 1: Write a simple reflex vacuum agent

Three rules cover every percept. That is all a simple reflex agent is.

```python
def reflex_agent(location, status):
    if status == "Dirty":
        return "Suck"
    if location == "A":
        return "Right"
    return "Left"
```

```python
check([("reflex_agent('A', 'Dirty')", 'Suck'), ("reflex_agent('A', 'Clean')", 'Right'), ("reflex_agent('B', 'Clean')", 'Left'), ("reflex_agent('B', 'Dirty')", 'Suck')])
```

### Solution to task 2: Size of a table driven agent

The numbers explode, which is the whole point: good agents are small programs, not giant tables.

```python
def table_size(percepts, lifetime):
    total = 0
    for t in range(1, lifetime + 1):
        total = total + percepts ** t
    return total
```

```python
check([('table_size(4, 3)', 84), ('table_size(4, 10)', 1398100), ('table_size(2, 1)', 2)])
```
