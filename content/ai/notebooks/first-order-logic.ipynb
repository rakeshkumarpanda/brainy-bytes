## Setup

```python
# A small helper. check(tests) runs each test of a task and tells you which ones pass.
# Each test is a pair: the call to make, and the answer we expect.
import json

def check(tests):
    passed = 0
    for call, expected in tests:
        try:
            got = eval(call)
        except Exception as error:
            got = "error: " + str(error)
        same = json.dumps(got, sort_keys=True, default=repr) == json.dumps(expected, sort_keys=True, default=repr)
        if same:
            passed = passed + 1
            print("PASS ", call)
        else:
            print("FAIL ", call, " expected", expected, " but got", got)
    if passed == len(tests):
        print("All", len(tests), "tests pass. Well done!")
    else:
        print(passed, "of", len(tests), "tests pass. Keep going, or look at the solution below.")
```

## Examples from the chapter

### Quantified sentences in a small model

```python
# A small model: the objects and which relations hold. Then check quantified sentences by looping over objects.
objects = ["Richard", "John", "TheCrown"]
person = {"Richard", "John"}
king = {"John"}
crown = {"TheCrown"}
on_head = {("TheCrown", "John")}
brother = {("Richard", "John"), ("John", "Richard")}

# for all x: King(x) => Person(x)
print("all kings are persons:", all((x not in king) or (x in person) for x in objects))
# for all x: King(x) and Person(x)   (the common mistake)
print("everything is a king and a person:", all((x in king) and (x in person) for x in objects))
# exists x: Crown(x) and OnHead(x, John)
print("a crown is on John's head:", any((x in crown) and ((x, "John") in on_head) for x in objects))
# for all x: Person(x) => exists y: Brother(x, y)
print("every person has a brother:", all((x not in person) or any((x, y) in brother for y in objects) for x in objects))
```

### Kinship relations from Parent facts

```python
# The kinship domain: define new relations from Parent with rules, then use them.
parent = [("Elizabeth", "Charles"), ("Philip", "Charles"), ("Elizabeth", "Anne"),
          ("Charles", "William"), ("Charles", "Harry")]

def children(p):
    return [c for a, c in parent if a == p]

def grandparents(x):
    # GrandParent(g, x) <=> exists p: Parent(g, p) and Parent(p, x)
    out = []
    for g, p in parent:
        if (p, x) in parent and g not in out:
            out.append(g)
    return out

def siblings(x):
    # Sibling(x, y) <=> x != y and exists p: Parent(p, x) and Parent(p, y)
    out = []
    for p, c in parent:
        if c == x:
            for s in children(p):
                if s != x and s not in out:
                    out.append(s)
    return out

print("grandparents of William:", grandparents("William"))
print("siblings of Charles:", siblings("Charles"))
print("siblings of Harry:", siblings("Harry"))
```

### Solution to task 1: Check quantified sentences in a model

The last test is an empty domain: there is no counterexample, so the ∀ sentence is true. (Standard FOL requires a non empty domain, but the code still has to give an answer.)

```python
def forall_implies(objects, p, q):
    for x in objects:
        if x in p and x not in q:
            return False
    return True

def exists_and(objects, p, q):
    for x in objects:
        if x in p and x in q:
            return True
    return False
```

```python
check([("forall_implies(['Richard', 'John', 'TheCrown'], {'John'}, {'Richard', 'John'})", True), ("forall_implies(['Richard', 'John'], {'Richard', 'John'}, {'John'})", False), ("exists_and(['Richard', 'John', 'TheCrown'], {'TheCrown'}, {'TheCrown'})", True), ("exists_and(['Richard', 'John'], {'Richard'}, {'John'})", False), ("forall_implies([], {'A'}, set())", True)])
```

### Solution to task 2: Kinship rules from Parent facts

Test 3 checks that Charles is listed once, even though he shares two parents with Anne.

```python
def grandparents(parent, x):
    out = []
    for g, p in parent:
        if (p, x) in parent and g not in out:
            out.append(g)
    return sorted(out)

def siblings(parent, x):
    out = []
    for p, c in parent:
        if c == x:
            for p2, other in parent:
                if p2 == p and other != x and other not in out:
                    out.append(other)
    return sorted(out)
```

### Solution to task 3: Count interpretations

With d = 1 every constant names the only object, and each predicate is either true or false of it: 1 × 2² × 2² = 16.

```python
def count_interpretations(d, constants, unary, binary):
    total = d ** constants
    total = total * (2 ** d) ** unary
    total = total * (2 ** (d * d)) ** binary
    return total
```

```python
check([('count_interpretations(2, 2, 0, 1)', 64), ('count_interpretations(3, 0, 1, 1)', 4096), ('count_interpretations(1, 3, 2, 2)', 16), ('count_interpretations(3, 1, 0, 0)', 3)])
```
