```python
# A small helper. check(tests) runs each test of a task and tells you which ones pass.
# Each test is a pair: the call to make, and the answer we expect.
import json

def check(tests):
    passed = 0
    for call, expected in tests:
        try:
            got = eval(call)
        except Exception as error:
            got = "error: " + str(error)
        same = json.dumps(got, sort_keys=True, default=repr) == json.dumps(expected, sort_keys=True, default=repr)
        if same:
            passed = passed + 1
            print("PASS ", call)
        else:
            print("FAIL ", call, " expected", expected, " but got", got)
    if passed == len(tests):
        print("All", len(tests), "tests pass. Well done!")
    else:
        print(passed, "of", len(tests), "tests pass. Keep going, or look at the solution below.")
```

## Examples from the chapter

### Depth first search on the slide tree

```python
# Depth first search on the slide tree: A has children B and C, B has D and E, and so on down to O.
tree = {"A": ["B", "C"], "B": ["D", "E"], "C": ["F", "G"], "D": ["H", "I"], "E": ["J", "K"],
        "F": ["L", "M"], "G": ["N", "O"]}

def dfs(node, goal, path):
    path = path + [node]
    print("visit", node)
    if node == goal:
        return path
    for child in tree.get(node, []):     # leaves have no entry, so they get []
        found = dfs(child, goal, path)
        if found is not None:
            return found
    return None                           # dead end: go back up (backtrack)

print("path:", dfs("A", "M", []))
```

### Iterative deepening on the slide tree

```python
# Iterative deepening: depth limited search with limit 0, then 1, then 2, ... until the goal is found.
tree = {"A": ["B", "C"], "B": ["D", "E"], "C": ["F", "G"], "D": ["H", "I"], "E": ["J", "K"],
        "F": ["L", "M"], "G": ["N", "O"]}

def dls(node, goal, limit):
    if node == goal:
        return [node]
    if limit == 0:
        return None                       # cut off: do not go deeper
    for child in tree.get(node, []):
        found = dls(child, goal, limit - 1)
        if found is not None:
            return [node] + found
    return None

for limit in range(5):
    result = dls("A", "M", limit)
    print("limit", limit, "->", result)
    if result is not None:
        break
```

### Solution to task 1: Depth limited search

DFS with a counter. The last test shows DFS returns the first goal it meets (via B), not the shortest (A, C).

```python
def dls(graph, node, goal, limit):
    if node == goal:
        return [node]
    if limit == 0:
        return None
    for child in graph.get(node, []):
        found = dls(graph, child, goal, limit - 1)
        if found is not None:
            return [node] + found
    return None
```

### Solution to task 2: Count the nodes of iterative deepening

The ratio ids_nodes ⁄ bfs_nodes tends to b ⁄ (b − 1).

```python
def ids_nodes(b, d):
    total = 0
    for i in range(d + 1):
        total = total + (d + 1 - i) * b ** i
    return total

def bfs_nodes(b, d):
    total = 0
    for i in range(d + 1):
        total = total + b ** i
    return total
```
